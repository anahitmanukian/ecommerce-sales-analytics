```python
import pandas as pd
from pathlib import Path

# 1. Automatically get project root directory and create target folder
project_dir = Path.cwd()
output_folder = project_dir / "data" / "clean"
output_folder.mkdir(parents=True, exist_ok=True)

# 2. Define path to Excel file
excel_file = project_dir / "E_commerce_cleaned.xlsx"

# 3. Validation columns to remove before exporting
columns_to_drop = {
    "CLEAN_orders": ["customer_check"],
    "CLEAN_order_items": ["order_check", "product_check"],
    "CLEAN_returns": ["order_check"],
}

# 4. Read sheets and export CLEAN_ tables to CSV
xls = pd.ExcelFile(excel_file)

for sheet_name in xls.sheet_names:
    if sheet_name.startswith("CLEAN_"):
        df = pd.read_excel(xls, sheet_name=sheet_name)

        # Remove temporary validation columns
        if sheet_name in columns_to_drop:
            df = df.drop(columns=columns_to_drop[sheet_name], errors="ignore")

        # Save to CSV
        csv_path = output_folder / f"{sheet_name}.csv"
        df.to_csv(csv_path, index=False)

        print(f"Saved: {csv_path}")

print("\nAll clean tables exported successfully!")
```

```python
# Clean whitespace to avoid subtle string mismatches
df_customers["customer_id"] = df_customers["customer_id"].astype(str).str.strip()
df_products["product_id"] = df_products["product_id"].astype(str).str.strip()
df_orders["order_id"] = df_orders["order_id"].astype(str).str.strip()
df_orders["customer_id"] = df_orders["customer_id"].astype(str).str.strip()
df_order_items["order_id"] = df_order_items["order_id"].astype(str).str.strip()
df_order_items["product_id"] = df_order_items["product_id"].astype(str).str.strip()
df_returns["order_id"] = df_returns["order_id"].astype(str).str.strip()

# Create lookup sets of valid parent IDs
valid_cust_ids = set(df_customers["customer_id"])
valid_prod_ids = set(df_products["product_id"])

# 1. Filter orders against customers
df_orders_clean = df_orders[
    df_orders["customer_id"].isin(valid_cust_ids)
].copy()
valid_order_ids = set(df_orders_clean["order_id"])

# 2. Filter order_items against BOTH valid orders AND valid products
df_items_clean = df_order_items[
    df_order_items["order_id"].isin(valid_order_ids)
    & df_order_items["product_id"].isin(valid_prod_ids)
].copy()

# 3. Filter returns against valid orders
df_returns_clean = df_returns[
    df_returns["order_id"].isin(valid_order_ids)
].copy()

# Overwrite CSVs
df_orders_clean.to_csv("data/clean/CLEAN_orders.csv", index=False)
df_items_clean.to_csv("data/clean/CLEAN_order_items.csv", index=False)
df_returns_clean.to_csv("data/clean/CLEAN_returns.csv", index=False)

print(f"Dropped {len(df_order_items) - len(df_items_clean)} orphan order items.")
print(f"Dropped {len(df_returns) - len(df_returns_clean)} orphan returns.")
```
